####  Nhập vào số nguyên dương n, tính tổng các chữ số của n 


```python
n = int(input())
tong = 0
m = 0
while n > 0:
    m = n%10
    tong += m
    n //= 10
print(tong)
```

####  Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ

```python
n = int(input())
demchan = 0
demle = 0
while n > 0:
    if n%2 == 0:
        demchan += 1
    else:
        demle += 1
    n //= 10
print(demchan, demle)
```

#### Nhập vào A, tìm n nhỏ nhất sao cho
#### 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 


```python
a = float(input())
n = 0
s = 0
while s <= a:
    n += 1
    s += 1/n
print(n)
```

#### Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1.
#### Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập 


```python
n = int(input())
nmax = n
nmin = n
while n != -1:
    n = int(input())
    if n > nmax:
        nmax = n
    if n < nmin:
        nmin = n
print(nmax, nmin)
    
```

#### [Fibonaci1]
#### Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ
#### bằng tổng hai số trước đó
#### Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt
#### quá A

```python
a = int(input())
t = 1
k = 1
while k <= a:
    t = k
    k = t + k
print(t)
```

#### Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có
#### ký tự đặc biệt, không số, không có dấu câu, chỉ có ký tự chữ và khoảng trắng)
#### Dùng cách: for trên index; KHÔNG dùng hàm split(), len()


```python
n = input()
dodai = 0
for i in n:
    dodai += 1
dem = 0
for i in range(dodai):
    if n[i] != " ":
        if i == 0 or n[i-1] == " ":
            dem+=1
print(dem)
```

#### Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi
#### Dùng cách: for trên index; KHÔNG dùng hàm split() 

```python
n = input()
a = []
for i in range(len(n)):
    if n[i] != " ":
        a += n[i]
    elif a:
        break
print(*a, sep ="")
```

#### Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành
#### thường và thường thành hoa.

```python
n = input()
a = ""
for i in n:
    if i.isupper():
        i = i.lower()
        a += i
    else:
        i = i.upper()
        a += i
print(a)
```

#### Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in
#### thường, bao nhiêu ký tự số 

```python
n = input()
demhoa = 0
demthg = 0
demso = 0
for i in n:
    if i.isupper():
        demhoa+=1
    elif i.islower():
        demthg+=1
    elif i.isdigit():
        demso+=1
print(demhoa,demthg,demso)
```
